## Menentukan Pertanyaan Bisnis

**SMART Question** adalah sebuah framework untuk merumuskan pertanyaan secara terstruktur agar memeperoleh informasi yang mendalam.

* **Spesific (Spesifik)**
  * Pertanyaan harus jelas, fokus pada sebuah topik tertentu, dan tidak bermakna ganda. Hindari pertanyaan yang terlalu luas.
    * Salah: Bagaimana cara meningkatkan penjualan?
    * Benar: Faktor apa saja yang memengaruhi penurunan penjualan produk kategori elektronik di wilayah Jakarta selama kuartal terakhir?
* **Measurable (Terukur)**
  * Pertanyaan harus bisa dijawab dengan angka atau matrix yang konkret, harus tahu apa yang akan dihitung.
    * Salah: Apakah pelanggan senang dengan layanan kita?
    * Benar: Berapa skor rata-rata Customer Satisfaction untuk layanan purna jual bulan ini dibandingkan bulan lalu?
* **Action-Oriented (Berorientasi Aksi)**
  * Hasil dari pertanyaan harus bisa memberikan arahan untuk melakukan tindakan nyata. Jika pertanyaan terjawab, stakeholder harus tahu apa langkah selanjutnya.
    * Salah: Mengapa orang suka berbelanja?
    * Benar: Fitur apa pada aplikasi yang paling sering digunakan sebelum pengguna memutuskan untuk melakukan checkout?
* **Relevant (Relevan)**
  * Hasil dari pertanyaan harus sejalan dengan tujuan utama bisnis atau masalah yang sedang dihadapi.
    * Salah: Menanyakan tentang stok gudang saat masalah utamanya adalah efektivitas kampanye media sosial.
    * Benar: Apakah kampanye iklan di Instagram memberikan Return on Ad Spend (ROAS) yang lebih tinggi dibandingkan iklan di TikTok?
* **Time-bound (Terikat Waktu)**
  * Pertanyaan harus ada batasan waktu yang jelas agar analisis memiliki konteks yang tepat.
    * Salah: Berapa banyak pengguna baru kita?
    * Benar: Berapa tingkat pertumbuhan pengguna baru secara bulanan (Month-over-Month) sepanjang tahun 2025?

**Steps to Take:**

- Mengisi *missing values* pada kolom kategorikal arah angin (`wd`) menggunakan nilai modus per stasiun.
- Menggabungkan kolom `year`, `month`, `day`, dan `hour` menjadi satu kolom tunggal `datetime` dengan tipe data `datetime64[ns]`.

- **Tipe Data:** Variabel penanda waktu masih terpecah dalam 4 kolom integer (`year`, `month`, `day`, `hour`), belum berupa objek `datetime` terpadu untuk analisis *time series*.
- **Sebaran Nilai:** Seluruh konsentrasi polutan bernilai non-negatif ($\ge 0$), tidak ditemukan nilai anomali negatif. Parameter cuaca berada pada rentang batas fisik wajar wilayah Beijing.

### Cleaning Data

#### Fixing Missing Values and Data Type Inconsistencies

```python
# Buat salinan DataFrame untuk proses cleaning
air_clean_df = air_df.copy()

# Gabungkan year, month, day, hour menjadi satu kolom datetime
air_clean_df['datetime'] = pd.to_datetime(air_clean_df[['year', 'month', 'day', 'hour']])

# Hapus kolom 'No' yang tidak memiliki nilai analisis
if 'No' in air_clean_df.columns:
  air_clean_df.drop(columns=['No'], inplace=True)

# Imputasi kolom numerik dengan interpolasi linear per stasiun
nums_cols = ['PM2.5', 'PM10', 'SO2', 'NO2', 'CO', 'O3', 'TEMP', 'PRES', 'DEWP', 'RAIN', 'WSPM']
for col in nums_cols:
   air_clean_df[col] = air_clean_df.groupby('station')[col].transform(
        lambda group: group.interpolate(method='linear').bfill().ffill()
    )

# Imputasi kolom kategorikal
air_clean_df['wd'] = air_clean_df.groupby('station')['wd'].transform(
    lambda group: group.bfill().ffill()
)

print(f"Total missing values tersisa: {air_clean_df.isna().sum().sum()}")
print(f"Bentuk data akhir: {air_clean_df.shape}")
print(f"Tipe data datetime: {air_clean_df['datetime'].dtype}")
air_clean_df.head()
```

**Insight:**
- **Penanganan Missing Values:** Seluruh *missing values* pada 11 fitur numerik (polutan dan meteorologi) berhasil ditangani secara optimal menggunakan teknik interpolasi linear per stasiun, menjaga kontinuitas data deret waktu (*time series*) tanpa mengubah distribusi asli. Fitur kategorikal `wd` diisi menggunakan propagasi nilai terdekat (*backward/forward fill*). Sisa *missing value* menjadi **0**.
- **Standarisasi Fitur Waktu:** Berhasil membentuk fitur tunggal `datetime` bertipe `datetime64[ns]` yang merepresentasikan runtutan waktu pengamatan per jam dari 1 Maret 2013 hingga 28 Februari 2017.
- **Reduksi Dimensi Redundan:** Kolom indeks `No` telah dihapus. DataFrame akhir memiliki **420.768 baris dan 18 kolom** dalam kondisi bersih, valid, dan siap untuk tahap analisis eksploratif (EDA).

## Exploratory Data Analysis (EDA)

### Explore Pertanyaan 1: Karakteristik Temporal dan Siklus Musiman Konsentrasi PM2.5

```python
# 1. Rangkuman statistik deskriptif polutan PM2.5 secara global
print("=== STATISTIK DESKRIPTIF PM2.5 (µg/m³) ===")
print(air_clean_df['PM2.5'].describe().round(2))

# 2. Agregasi rata-rata dan median PM2.5 per tahun
yearly_pm25 = air_clean_df.groupby('year')['PM2.5'].agg(['mean', 'median', 'min', 'max', 'std']).round(2)
print("\n=== KONSENTRASI PM2.5 PER TAHUN ===")
print(yearly_pm25)

# 3. Agregasi rata-rata PM2.5 berdasarkan bulan (siklus tahunan)
monthly_pm25 = air_clean_df.groupby('month')['PM2.5'].agg(['mean', 'median', 'std']).round(2)
print("\n=== KONSENTRASI PM2.5 BERDASARKAN BULAN (1 = Jan, 12 = Des) ===")
print(monthly_pm25.sort_values(by='mean', ascending=False))

# 4. Agregasi rata-rata PM2.5 per stasiun
station_pm25 = air_clean_df.groupby('station')['PM2.5'].agg(['mean', 'median', 'std']).round(2)
print("\n=== RATA-RATA PM2.5 PER STASIUN (DIURUTKAN DARI TERTINGGI) ===")
print(station_pm25.sort_values(by='mean', ascending=False))
```

### Explore Pertanyaan 2: Pengaruh Parameter Meteorologi terhadap Polutan PM2.5

```python
# 1. Matriks korelasi parameter cuaca terhadap seluruh polutan
weather_cols = ['TEMP', 'PRES', 'DEWP', 'RAIN', 'WSPM']
pollutant_cols = ['PM2.5', 'PM10', 'SO2', 'NO2', 'CO', 'O3']

# Hitung korelasi khusus terhadap PM2.5
pm25_corr = air_clean_df[weather_cols + ['PM2.5']].corr()['PM2.5'].drop('PM2.5').sort_values()
print("=== KOEFISIEN KORELASI PARAMETER CUACA DENGAN PM2.5 ===")
print(pm25_corr.round(3))

# 2. Analisis rata-rata PM2.5 berdasarkan kondisi hujan vs tidak hujan
air_clean_df['is_rain'] = np.where(air_clean_df['RAIN'] > 0, 'Hujan', 'Tidak Hujan')
rain_comparison = air_clean_df.groupby('is_rain')['PM2.5'].agg(['mean', 'median', 'count']).round(2)
print("\n=== PERBANDINGAN RATA-RATA PM2.5: HUJAN VS TIDAK HUJAN ===")
print(rain_comparison)

# 3. Analisis rata-rata PM2.5 berdasarkan kategori kecepatan angin
air_clean_df['wind_category'] = pd.cut(
    air_clean_df['WSPM'],
    bins=[-np.inf, 1.5, 3.0, np.inf],
    labels=['Tenang (< 1.5 m/s)', 'Sedang (1.5 - 3.0 m/s)', 'Kencang (> 3.0 m/s)']
)
wind_comparison = air_clean_df.groupby('wind_category')['PM2.5'].agg(['mean', 'median', 'count']).round(2)
print("\n=== PERBANDINGAN RATA-RATA PM2.5 BERDASARKAN KECEPATAN ANGIN ===")
print(wind_comparison)
```

**Insight Eksplorasi Pertanyaan 2:**
- **Efek Dispersi Angin (`WSPM`, r = -0.27):** Kecepatan angin memiliki korelasi negatif terkuat terhadap PM2.5. Saat angin tenang (< 1.5 m/s), rata-rata PM2.5 melonjak menjadi **97.06 µg/m³**. Saat angin bertiup kencang (> 3.0 m/s), rata-rata PM2.5 terpangkas drastis menjadi **35.45 µg/m³**.
- **Efek Kelembapan & Kondensasi (`DEWP`, r = 0.26):** Titik embun berhubungan positif signifikan dengan PM2.5, menunjukkan kelembapan udara mempercepat reaksi kimia pembentukan partikel polutan sekunder.
- **Efek Presipitasi (`RAIN`):** Saat terjadi hujan, rata-rata PM2.5 turun dari **80.56 µg/m³** menjadi **62.34 µg/m³**, membuktikan air hujan efektif mencuci partikel tersuspensi (*wet scavenging*).

### Explore Pertanyaan 3 (Analisis Lanjutan): Klasifikasi Kualitas Udara (Binning AQI) dan Identifikasi Wilayah Kritis

```python
# 1. Klasifikasi PM2.5 ke dalam kategori standar US-EPA / WHO menggunakan pd.cut
bins = [-np.inf, 12, 35, 55, 150, 250, np.inf]
labels = ['Good', 'Moderate', 'Unhealthy for Sensitive', 'Unhealthy', 'Very Unhealthy', 'Hazardous']
air_clean_df['AQI_Category'] = pd.cut(air_clean_df['PM2.5'], bins=bins, labels=labels)

# 2. Distribusi frekuensi dan persentase kategori AQI global
aqi_counts = air_clean_df['AQI_Category'].value_counts()
aqi_percent = (air_clean_df['AQI_Category'].value_counts(normalize=True) * 100).round(2)
aqi_summary = pd.DataFrame({'Jumlah Jam': aqi_counts, 'Persentase (%)': aqi_percent}).reindex(labels)
print("=== DISTRIBUSI KATEGORI KUALITAS UDARA (AQI) KESELURUHAN ===")
print(aqi_summary)

# 3. Proporsi kondisi kritis (Very Unhealthy + Hazardous: PM2.5 > 150 µg/m³) per stasiun
critical_hours = air_clean_df[air_clean_df['AQI_Category'].isin(['Very Unhealthy', 'Hazardous'])]
station_critical_prop = (
    critical_hours.groupby('station').size() / air_clean_df.groupby('station').size() * 100
).sort_values(ascending=False).round(2)

print("\n=== PROPORSI JAM KATEGORI KRITIS (PM2.5 > 150 µg/m³) PER STASIUN (%) ===")
print(station_critical_prop)
```

**Insight Eksplorasi Pertanyaan 3 (Berdasarkan Data Aktual):**

- **Dominasi Kategori Tidak Sehat:** Dari total 420.768 jam pengamatan, kualitas udara didominasi oleh kategori **Unhealthy** sebesar **34.57% (145.462 jam)**. Jika diakumulasikan, udara tidak sehat (*Unhealthy for Sensitive* hingga *Hazardous*) mencakup **62.70%** dari keseluruhan waktu pengamatan. Udara kategori **Good** hanya dinikmati warga selama **15.19% waktu (63.928 jam)**.
- **Kondisi Bahaya Ekstrem:** Kondisi polusi level kritis (*Very Unhealthy* dan *Hazardous*, PM2.5 > 150 µg/m³) terjadi selama **15.18% waktu (63.860 jam)**, membuktikan tingginya frekuensi darurat polusi akut di wilayah metropolitan Beijing.

## Analisis Lanjutan (Opsional)

```python
# 1. Menyiapkan data koordinat estimasi stasiun (Latitude & Longitude wilayah Beijing)
station_locations = {
    'station': ['Aotizhongxin', 'Changping', 'Dingling', 'Dongsi', 'Guanyuan', 'Gucheng',
                'Huairou', 'Nongzhanguan', 'Shunyi', 'Tiantan', 'Wanliu', 'Wanshouxigong'],
    'lat': [39.982, 40.217, 40.292, 39.929, 39.929, 39.914, 40.328, 39.933, 40.127, 39.886, 39.987, 39.878],
    'lon': [116.397, 116.230, 116.220, 116.417, 116.339, 116.184, 116.628, 116.460, 116.655, 116.407, 116.287, 116.352]
}
geo_df = pd.DataFrame(station_locations)

# 2. Gabungkan dengan rata-rata PM2.5 per stasiun
station_stats = air_clean_df.groupby('station')['PM2.5'].mean().reset_index()
geo_data = pd.merge(geo_df, station_stats, on='station')

# 3. Membuat Peta Interaktif dengan Folium
# Pakai tile Esri WorldStreetMap (bebas blokir & tanpa API key)
beijing_map = folium.Map(
    location=[40.0, 116.4],
    zoom_start=9,
    tiles='https://server.arcgisonline.com/ArcGIS/rest/services/World_Street_Map/MapServer/tile/{z}/{y}/{x}',
    attr='Esri'
)

for index, row in geo_data.iterrows():
    # Tentukan warna marker berdasarkan level polusi
    color = 'red' if row['PM2.5'] > 80 else 'orange' if row['PM2.5'] > 70 else 'green'

    folium.CircleMarker(
        location=[row['lat'], row['lon']],
        radius=row['PM2.5']/5, # Ukuran lingkaran proporsional terhadap polusi
        popup=f"Stasiun: {row['station']}<br>Rata-rata PM2.5: {row['PM2.5']:.2f} µg/m³",
        color=color,
        fill=True,
        fill_color=color,
        fill_opacity=0.6
    ).add_to(beijing_map)

# Tampilkan peta
beijing_map
```

**Insight:**
- **Visualisasi Spasial Polusi:** Peta interaktif menunjukkan bahwa titik dengan radius lingkaran terbesar dan warna merah pekat terpusat di area distrik bisnis dan pusat pemerintahan (tengah-selatan Beijing).
- **Efek Lokasi:** Terdapat degradasi tingkat polusi dari arah Selatan (lebih tercemar) menuju Utara (lebih bersih). Hal ini dipengaruhi oleh letak geografis stasiun di Utara yang berada dekat pegunungan, yang memberikan perlindungan alami dan sirkulasi udara yang lebih segar dibandingkan wilayah dataran rendah di Selatan yang padat pemukiman.
